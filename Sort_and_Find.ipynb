```python
import pandas as pd
from collections import Counter

# Read the CSV file into a Pandas DataFrame
df = pd.read_csv('MetaLatin.csv')

# Sort the DataFrame by the "title" column
df = df.sort_values(by='title')

# Extract the first 5 words of each entry in the "title" column
first_five_words = df['title'].fillna('').apply(lambda x: ' '.join(x.split()[:5]))

# Count the frequency of each phrase
phrase_freq = Counter(word for word in first_five_words if word != '')

# Get the top 10 most frequent phrases
most_common_phrases = phrase_freq.most_common(10)

# Print the results
print("Top 10 most frequent phrases in the first 5 words of each title:")
for phrase, freq in most_common_phrases:
    print(f"{phrase}: {freq}")
```

```python
new_df = pd.DataFrame()
for phrase, freq in most_common_phrases:
    phrase_df = df[first_five_words == phrase].head(20)
    new_df = pd.concat([new_df, phrase_df])

# Convert the new DataFrame to a CSV file
new_df.to_csv('Testing_Freq_Lat.csv', index=False)
```
