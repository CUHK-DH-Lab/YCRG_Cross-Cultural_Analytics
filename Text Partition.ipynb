```python
import os
```

```python
# Set the input file
input_file = r"E:\Datasets\Hathi Catalogue\hathi_full_20240301.txt"

# Set the number of lines per output file
lines_per_file = 1000

# Set the output file prefix
output_prefix = "HT_Part_"
```

```python
# Open the input file
with open(input_file, "r",encoding = 'cp1252') as f:
    # Initialize a counter for the number of lines read
    line_count = 0
    # Initialize a variable to hold the current output file
    output_file = None
    # Loop through the lines in the input file
    for line in f:
        # If we have reached the maximum number of lines for the current output file
        if line_count % lines_per_file == 0:
            # Close the current output file
            if output_file is not None:
                output_file.close()
            # Open a new output file with the next number in the sequence
            output_file = open(f"{output_prefix}{line_count // lines_per_file}.txt", "w")
        # Write the line to the current output file
        output_file.write(line)
        # Increment the line counter
        line_count += 1
    # Close the last output file
    if output_file is not None:
        output_file.close()
```
