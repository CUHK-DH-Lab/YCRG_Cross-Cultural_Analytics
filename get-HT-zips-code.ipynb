### Using `ht_text_prep` Python library to get zipfiles at end of pairtree folder structure
When downloading public domain full text data from HathiTrust, it comes in a format that is optimized for retrieval and transferring, but not necessarily for easily accessing files stored locally. This format is called "pairtree" where there is first a top-level folder for each submitting institution (such as `mdp` for U. Michigan) and then nested folders for every two numbers in the volume ID's numerical suffix. So, a file with HTID mdp.3456784679 would be stored at the path:

`mdp/34/56/78/46/79/mdp.3456784679.zip`

It can take a while to move all these files to be easily programmatically accessed, so we have a tool that helps with this called the `ht_text_prep` Python library (see: https://pypi.org/project/ht-text-prep/). This notebook will take you through using the library to move the zip files, but first we need to install it via `pip`:

```python
# importing libraries we'll be using
import pandas as pd
import ht_text_prep as htp
import glob
import os
from tqdm.notebook import tqdm
```

This code originally took data from three folders holding about 60k files each, and moved them to new, flat folders, rather than nested pairtree folders. In the next cell, we find the three top level folders and get their paths.

```python
# get current working directory
cwd = os.getcwd()
print(cwd)

# Top directory with the files downloaded from rsync
source_paths = glob.glob(r'E:\Datasets\HathiTrust_Rsync\ia\*')

# Segmenting for testing
# source_paths = source_paths[4:7]
print(f"Found {len(source_paths)} folders")
source_paths
```

Next, we pair our source paths with new output folders as tuples. Change the names below with the actual source paths followed by the output folders, each looking like below, and stored in a list:

`(source_path, output_path)`


```python
# change the strings below to match your source and output folder paths
dir_pairs = [(r'E:\Datasets\HathiTrust_Rsync\ia', r'E:\Datasets\HathiTrust_flatten\ia')]

```

Next, we use the `ht_text_prep` library to get the zips from each source folder path and move them to the output paths using a method `htp.getzips()` where the parameters are `source_path` and `output_path`

```python
# move files
for t in dir_pairs:
    data_dir = t[0]
    out_dir = t[1]
    count = len(glob.glob(f"{data_dir}/**", recursive=True))
    print(count)
    htp.get_zips(data_dir, out_dir)
```
